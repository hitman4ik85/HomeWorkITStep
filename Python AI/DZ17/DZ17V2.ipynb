# DZ17 — Нейронна мережа для розпізнавання чисел в інтервалі [5; 15]

У цьому завданні ми будуємо просту нейронну мережу на **PyTorch**, яка вчиться відрізняти числа:

- якщо число належить інтервалу від **5 до 15** включно — це **клас 1**;
- якщо число менше 5 або більше 15 — це **клас 0**.

Спочатку реалізуємо **базову модель** через `nn.Sequential`, а потім зробимо **ускладнену модель** у вигляді класу `Net`.
Після цього порівняємо навчання обох моделей та їхні лінії рішень.

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# Підготуємо навчальні дані
# Візьмемо числа на відрізку [0; 20] з рівномірним кроком
X = torch.linspace(0., 20., steps=100).unsqueeze(1)  # тензор розміру (100, 1)

# Створюємо мітки: 1, якщо число в [5; 15], інакше 0
y = ((X >= 5.) & (X <= 15.)).float()  # (100, 1)

X[:10], y[:10]
```

### Дослідження
- `torch.linspace(0., 20., steps=100)` генерує рівномірну сітку з 100 чисел від 0 до 20.
- `unsqueeze(1)` додає стовпчиковий вимір, щоб кожне число стало маленьким вектором розміру `1` — так модель очікує вхідні дані.
- Тензор `X` — це **вхідні ознаки** (features), а тензор `y` — це **цільові мітки** (labels).
- Мітки `y` обчислюємо через логічну умову: якщо `x` у проміжку `[5; 15]`, значення `True`, в іншому випадку — `False`.
- Після `.float()` отримаємо 0.0 або 1.0, що підходить для `BCELoss`, яка працює з дійсними значеннями ймовірностей і класів 0/1.


## Базова модель: `nn.Sequential`

Зробимо просту мережу через `Sequential`:
- прихований шар `Linear(1, 8)`;
- функція активації `ReLU`;
- вихідний шар `Linear(8, 1)`;
- `Sigmoid` на виході, щоб отримати ймовірність класу 1.

Такий варіант вже може сформувати «плато» для інтервалу `[5; 15]`, на відміну від моделі з одним нейроном.

```python
# Базова модель (Sequential)
model = nn.Sequential(
  nn.Linear(1, 8),  # 1 -> 8
  nn.ReLU(),        # нелінійність
  nn.Linear(8, 1),  # 8 -> 1
  nn.Sigmoid()      # ймовірність класу 1
)

loss_fn = nn.BCELoss()
optim = torch.optim.SGD(model.parameters(), lr=.1)
```

### Дослідження
- `nn.Linear(1, 8)` перетворює одне число у вектор з 8 ознак — це перший прихований шар.
- `ReLU` додає нелінійність, щоб модель могла побудувати складнішу форму залежності, ніж проста пряма лінія.
- `nn.Linear(8, 1)` зводить 8 прихованих ознак до одного числа (логіта), з якого далі обчислюється ймовірність.
- `Sigmoid` переводить логіт у число від 0 до 1 — це інтерпретуємо як ймовірність того, що число належить інтервалу `[5; 15]`.
- `BCELoss` працює з ймовірностями та мітками 0/1 і вимірює, наскільки передбачення моделі відрізняються від правильних відповідей.
- Оптимізатор `SGD` з відносно великою швидкістю навчання `0.1` крок за кроком змінює ваги шарів у напрямку зменшення помилки.


```python
# Навчання базової моделі
n_epochs = 2000
loss_history_basic = []

for epoch in range(n_epochs):
  y_pred = model(X)
  loss = loss_fn(y_pred, y)

  optim.zero_grad()
  loss.backward()
  optim.step()

  loss_history_basic.append(loss.item())

  if epoch % 200 == 0:
    print(f"Epoch: {epoch} | Loss: {loss.item():.4f}")

loss_history_basic[-1]
```

### Дослідження
- Цикл навчання побудований за класичною схемою: `forward -> loss -> zero_grad -> backward -> step`.
  1. на кожній епосі рахуємо передбачення `y_pred = model(X)`;
  2. обчислюємо помилку `loss = loss_fn(y_pred, y)`;
  3. `optim.zero_grad()` — обнуляємо застарілі градієнти, щоб вони не накопичувались;
  4. `loss.backward()` — автоматично рахуємо похідні (градієнти) за всіма параметрами моделі;
  5. `optim.step()` — оновлюємо ваги згідно з обраним оптимізатором.
- Значення `loss.item()` додаємо до списку `loss_history_basic`, щоб потім побачити, як помилка змінювалась з епохами.
- Якщо фінальне значення loss невелике, це означає, що модель навчилася добре відокремлювати числа всередині інтервалу від решти.


```python
# Графік зміни помилки під час навчання базової моделі
plt.plot(loss_history_basic)
plt.xlabel('Епоха')
plt.ylabel('Loss')
plt.title('Навчання базової моделі')
plt.show()
```

### Дослідження
- Крива помилки показує, як модель поступово навчається.
- Якщо графік спадає і виходить на плато, це хороша ознака.
- Якби помилка росла або хаотично стрибала, треба було б змінити архітектуру або швидкість навчання.

```python
# Тестування базової моделі на вигаданих даних
test = torch.tensor([[0.], [3.], [5.], [7.], [10.], [15.], [17.], [20.]], dtype=torch.float32)

with torch.no_grad():
  pred = model(test)

print("Базова модель:")
for i in range(len(test)):
  x = test[i]       # беремо i-те число
  p = pred[i]       # беремо i-ту ймовірність

  value = x.item()
  proba = p.item()
  label = int(proba >= 0.5)

  print(f"Число {value:4.1f} -> ймовірність класу 1 = {proba:.3f}, прогнозований клас = {label}")
```

### Дослідження
- Для тестування ми подаємо в модель конкретні числа `0, 3, 5, 7, 10, 15, 17, 20`.
- Для кожного числа виводимо:
  - саме значення `x`;
  - ймовірність класу 1 (належність до інтервалу `[5; 15]`);
  - прогнозований клас 0 або 1 при порозі 0.5.
- Такий формат виводу зручний, щоб "на око" перевірити логіку: числа всередині `[5; 15]` мають давати клас 1, поза інтервалом — клас 0.


```python
# Візуалізація лінії рішення базової моделі
with torch.no_grad():
  grid_x = torch.linspace(0., 20., steps=400).unsqueeze(1)
  grid_proba_basic = model(grid_x).squeeze(1)

plt.figure(figsize=(7, 4))
plt.plot(grid_x.numpy(), grid_proba_basic.numpy())
plt.axvspan(5, 15, color='green', alpha=.1, label='Інтервал [5; 15]')
plt.axhline(0.5, color='red', linestyle='--', label='Поріг 0.5')
plt.xlabel('x')
plt.ylabel('P(клас = 1)')
plt.title('Лінія рішення базової моделі')
plt.legend()
plt.show()
```

### Дослідження
- По осі `x` відкладаємо усі числа від 0 до 20 з кроком, а по осі `y` — ймовірність `P(клас = 1)`.
- Зелена напівпрозора зона показує справжній інтервал `5–15`, який ми вважаємо класом 1.
- Горизонтальна пунктирна лінія на рівні 0.5 — це поріг, за яким ми перетворюємо ймовірність на клас.
- Лінія моделі має бути високою (близько 1) всередині зеленої зони та низькою (близько 0) поза нею — тоді модель поводиться інтуїтивно правильно.


## Ускладнена модель: клас `Net` з кількома шарами

Тепер зробимо мережу складнішою:
- оголосимо власний клас `Net`, який наслідується від `nn.Module`;
- додамо ще один прихований шар;
- використаємо різні функції активації (`Tanh`, `ReLU`, `Sigmoid`).

```python
# Оголошення складнішої мережі
class Net(nn.Module):
  def __init__(self):
    super().__init__()
    self.layer1 = nn.Linear(1, 8)
    self.layer2 = nn.Linear(8, 8)
    self.layer3 = nn.Linear(8, 1)
    self.tanh = nn.Tanh()
    self.relu = nn.ReLU()
    self.sigmoid = nn.Sigmoid()

  def forward(self, X):
    X = self.layer1(X)
    X = self.tanh(X)
    X = self.layer2(X)
    X = self.relu(X)
    X = self.layer3(X)
    X = self.sigmoid(X)
    return X

model2 = Net()
loss_fn2 = nn.BCELoss()
optim2 = torch.optim.Adam(model2.parameters(), lr=.01)
```

### Дослідження
- У `Net` є три лінійні шари: `1 -> 8 -> 8 -> 1`, тобто ми поступово збільшуємо представлення, обробляємо його та стискаємо назад.
- Після першого шару використовуємо `Tanh`, як у NLP-прикладі, щоб отримати значення в діапазоні [-1; 1] і мати більш плавну активацію.
- Після другого шару використовуємо `ReLU`, як у задачах з комп'ютерним зором — вона добре працює для глибших мереж.
- `Sigmoid` на виході знову дає ймовірність класу 1, з якою можна працювати як у звичайній бінарній класифікації.
- Оптимізатор `Adam` автоматично підбирає ефективний крок оновлення для кожного параметра, тому часто сходиться швидше та стабільніше, ніж чистий `SGD`.


```python
# Навчання складнішої моделі
n_epochs2 = 2000
loss_history_deep = []

for epoch in range(n_epochs2):
  y_pred2 = model2(X)
  loss2 = loss_fn2(y_pred2, y)

  optim2.zero_grad()
  loss2.backward()
  optim2.step()

  loss_history_deep.append(loss2.item())

  if epoch % 200 == 0:
    print(f"[Net] Epoch: {epoch} | Loss: {loss2.item():.4f}")

loss_history_deep[-1]
```

### Дослідження
- Цикл навчання для `Net` повністю аналогічний базовій моделі, але тепер оновлюються ваги складнішої мережі.
- Список `loss_history_deep` дозволяє відстежувати, як помилка змінюється з епохами саме для ускладненої архітектури.
- Порівнюючи `loss_history_basic` і `loss_history_deep`, ми можемо зробити висновок, чи дає додаткова складність реальний виграш.


```python
# Графік навчання ускладненої моделі
plt.plot(loss_history_deep)
plt.xlabel('Епоха')
plt.ylabel('Loss')
plt.title('Навчання ускладненої моделі (Net)')
plt.show()

# Порівняння кривих навчання двох моделей на одному графіку
plt.plot(loss_history_basic, label='Базова модель (Sequential + SGD)')
plt.plot(loss_history_deep, label='Ускладнена модель (Net + Adam)')
plt.xlabel('Епоха')
plt.ylabel('Loss')
plt.title('Порівняння навчання моделей')
plt.legend()
plt.show()
```

### Дослідження
- Перший графік показує, як навчається тільки `Net` — можна оцінити, чи зменшується помилка та чи немає перенавчання.
- Другий графік дає можливість візуально порівняти, яка модель швидше і нижче зменшує помилку на тих самих даних.
- Якщо крива `Net` падає швидше або до нижчого рівня, це означає, що ускладнена архітектура + Adam краще підходять для даної задачі.
- Якщо ж криві дуже схожі, можна зробити висновок, що для такої простої задачі достатньо і базової моделі.


```python
# Тестування ускладненої моделі на тих самих даних
with torch.no_grad():
  pred2 = model2(test)

print("Ускладнена модель (Net):")
for i in range(len(test)):
  x = test[i]
  p = pred2[i]

  value = x.item()
  proba = p.item()
  label = int(proba >= 0.5)

  print(f"Число {value:4.1f} -> ймовірність класу 1 = {proba:.3f}, прогнозований клас = {label}")
```

### Дослідження
- Для чесного порівняння тестуємо `Net` на тих самих числах, що й базову модель.
- Формат виводу такий самий: число, ймовірність класу 1, прогнозований клас при порозі 0.5.
- Порівнюючи рядки для двох моделей, можна побачити, де ускладнена мережа більш "впевнена" (ймовірність ближча до 0 або 1).


```python
# Візуалізація лінії рішення ускладненої моделі
with torch.no_grad():
  grid_x = torch.linspace(0., 20., steps=400).unsqueeze(1)
  grid_proba_deep = model2(grid_x).squeeze(1)

plt.figure(figsize=(7, 4))
plt.plot(grid_x.numpy(), grid_proba_deep.numpy())
plt.axvspan(5, 15, color='green', alpha=.1, label='Інтервал [5; 15]')
plt.axhline(0.5, color='red', linestyle='--', label='Поріг 0.5')
plt.xlabel('x')
plt.ylabel('P(клас = 1)')
plt.title('Лінія рішення ускладненої моделі (Net)')
plt.legend()
plt.show()
```

### Дослідження
- Тут бачимо, як ускладнена модель розділяє числа на всьому відрізку `[0; 20]` так само, як ми робили для базової.
- Ідеально, якщо ймовірність близька до 1 в `[5; 15]` та близька до 0 поза цим інтервалом — тоді модель чітко фіксує кордони інтервалу.
- Якщо форма кривої відрізняється від базової моделі (наприклад, перехід різкіший або плавніший), можна обговорити, яка поведінка краще для задачі.


## Підсумок

- **Базова модель** на `Sequential` з одним прихованим шаром уже добре вчиться розрізняти числа в інтервалі `[5; 15]`.
- **Ускладнена модель** `Net` з додатковим шаром і активаціями `Tanh + ReLU` демонструє, як можна об'єднати ідеї з попередніх прикладів, які ми пройшли.
- Графіки навчання та ліній рішень дозволяють візуально порівняти, як кожна модель поводиться на даних.
- Важливо не лише отримати низький loss, але й подивитися на **лінію рішення** та якість передбачень на тестових прикладах.
- Такий підхід відповідає "науковому міні-дослідженню": ми пробуємо різні архітектури та аналізуємо, як вони вчаться визначати числа від 5 до 15.
