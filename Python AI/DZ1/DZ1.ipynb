**7. Чому дорівнює значення змінної `let`?**

```python
arr = [1, -2, 5, 12]
let = sum(arr[1:])
let
```

**Дослідження:**  
- Маємо список(лист) з числами. Вираз `arr[1:]` означає зріз з індексу 1 і до кінця списку, тобто беремо всі елементи **починаючи з другого**.  
- Функція `sum(...)` обчислює суму чисел у цьому зрізі. У нашому випадку: `[-2, 5, 12]`.  
- Отже, `sum([-2, 5, 12]) = 15`, тому змінна `let` дорівнює **15**.

**8. Яке значення змінної `word`?**

```python
obj = ['coral', 'marlin', 'in', 'Australia']
word = max(obj, key=len)
word
```

**Дослідження:**  
- Маємо список(лист) з рядків. Вбудована функція `max(iterable, key=...)` повертає елемент з **найбільшим значенням ключа**.  
- Тут `key=len` — ключем є довжина рядка. Отже, повертається рядок з **найбільшою довжиною**.  
- У нашому випадку найдовше слово — `'Australia'`, тому `word` дорівнює **'Australia'**.

**9. Який результат фрагменту коду?**

```python
def off(a: int, b: int) -> int:
    return b ** a

off(2, 5)
```

**Дослідження:**  
- Описано функцію `off(a, b)`, яка повертає `b ** a`. Оператор `**` — це піднесення до степеня.  
- Тобто ми підносимо `b` у степінь `a`. Для `off(2, 5)` це означає `5 ** 2 = 25`.  
- Отже, результат виклику `off(2, 5)` дорівнює **25**.

**10. Чому дорівнює значення змінної `hello`?**

```python
hello = filter(lambda i: i % 2 == 0, [8, -12, 5])
list(hello)
```

**Дослідження:**  
- Використано `filter(функція, послідовність)`. Тут `lambda i: i % 2 == 0` — це умова **парності**: число ділиться на 2 без остачі.  
- Фільтруємо список `[8, -12, 5]` і лишаємо тільки парні елементи. У Python 3 `filter(...)` повертає ітератор, тому для перегляду значень використовуємо `list(hello)`.  
- У нашому випадку парні — це `8` та `-12`, тому результат — **`[8, -12]`**.
