**Завдання:** Якщо результат `TRUE` — використовуємо 🍉, якщо `FALSE` — 🍪.

```python
all([True, True, False, True])
```

**Відповідь:** 🍪 *(False)*

**Дослідження:**  
- All, це є всі значення(елементи). All приймає на себе список(лист) з елементами Правда та Хибність (True or False), з логічним типом даних (bool). 
- Тобто: Всі елементи того, що ми передаємо в вбудовану функцію All повинні бути True, тільки тоді повертається True, в іншому випадку, якщо у списку є хочаб один False, то одразу йде ствердження, що у цьому випадку відповідь буде False.

```python
all([12 % 3 == 0, 2 ** 3 == 8, 9 // 2 == 4])
```

**Відповідь:** 🍉 *(True)*

**Дослідження:**  
- All перевіряє всі умови в списку(листі) на істинність (True).  
- У нас: `12 % 3 == 0` це True, `2 ** 3 == 8` це True, `9 // 2 == 4` це теж True (бо ціле ділення 9 на 2 дорівнює 4).  
- Тобто всі елементи є True, отже All повертає True.

```python
any([8 ** 2 != 64, len('climate') == 8, type('8') is str])
```

**Відповідь:** 🍉 *(True)*

**Дослідження:**  
- Any означає будь-що(хочаб один). Any приймає список(лист) з булевих значень і повертає True, якщо **хочаб один** елемент є True.  
- У нас: `8 ** 2 != 64` це False, `len('climate') == 8` це False (бо довжина слова ‘climate’ дорівнює 7), `type('8') is str` це True.  
- Є хочаб один True, отже Any повертає True.

```python
collection = [2, 2.01, '2.001']
all(type(i) is str for i in collection)
```

**Відповідь:** 🍪 *(False)*

**Дослідження:**  
- All перевіряє, що **всі** елементи генератора є True. Вираз `type(i) is str` повертає True тільки для рядків.  
- У списку є типи: число(ціле) `int`, число(дробове) `float` та рядок `str`. Не всі елементи є рядками, отже буде хочаб один False і All поверне False.

```python
dictionary = {'+': 8, '*': 0, '~': 2}
if all(value >= 0 for value in dictionary.values()):
    print('🍉')
else:
    print('🍪')
```

**Відповідь:** 🍉 *(True)*

**Дослідження:**  
- All перевіряє всі значення словника через умову `value >= 0`. У нас значення: 8, 0, 2 — всі не від’ємні (>= 0), отже всі перевірки є True.  
- Тому All повертає True і виводиться 🍉.
