**Завдання:** 
- Дано матрицю NxN, яка у випадковому порядку має змогу містити `🍪`. 
- Написати програму яка повертає матрицю, і повертає адресу розташування `🍪`.

```python
def build_matrix(n, cookie_i=None, cookie_j=None, icon='🍪'):
    matrix = [['~'] * n for _ in range(n)]
    if cookie_i is not None and cookie_j is not None:
        matrix[cookie_i][cookie_j] = icon
    return matrix

def print_matrix(matrix):
    for row in matrix:
        print(" ".join(f"\t{element}" for element in row))

def find_cookie(matrix, icon='🍪'):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            if matrix[i][j] == icon:
                print(f"Знайдено іконку '{icon}' за адресою: row={i}, col={j}")
                print(f"Візуально: рядок={i+1}, стовбець={j+1}")
                return (i, j)
    print("Іконку не знайдено")
    return None
```

**Дослідження:**  
1) `build_matrix(n, i, j, icon)` створює матрицю `n×n` із символів `'~'`. Якщо передані `i` і `j`, то в цю клітинку ставиться іконка `icon` (у нашому випадку `'🍪'`). Індексація тут 0-базова (починаємо з 0).  
2) `print_matrix(matrix)` друкує матрицю построчно, між елементами використовується табуляція `\t`, щоб було зручно дивитись.  
3) `find_cookie(matrix)` проходить подвійним циклом по всій матриці: спочатку всі рядки `i`, всередині — всі стовпці `j`. Якщо знаходить іконку, то одразу друкує адресу у 0-базовій формі (`row=i, col=j`) та додатково **Візуально** у 1-базовій формі (`рядок=i+1, стовбець=j+1`). Якщо іконки немає — друкується повідомлення.  
Подвійний цикл перевіряє всі клітинки `n×n`, тому складність алгоритму — `O(n²)`, де `O` це порядок зростання.
Чому саме `n²` - на кожну ітерацію зовнішнього, внутрішній теж іде `n` разів. Разом: `n * n = n²` базових операцій → `O(n²)`.

**Демонстрація: 3×3 з іконкою в (1, 2) — 0-базово (тобто візуально це 2-й рядок, 3-й стовпець)**

```python

n = 3
matrix = build_matrix(n, cookie_i=1, cookie_j=2, icon='🍪')

print("Матриця:")
print_matrix(matrix)

address = find_cookie(matrix)
address
```

**Дослідження:**  
1. Ми створили матрицю 3×3 та поставили іконку у позицію `(1, 2)` у 0-базовій індексації. Це означає **Візуально** 2-й рядок і 3-й стовпець.  
2. Функція друкує обидві адреси (0-базову та **Візуально** 1-базову) і повертає кортеж `address = (1, 2)` як 0-базову адресу.

**Приклад з випадковим розташуванням печива:**

```python
import random

n2 = 4
ri = random.randint(0, n2 - 1)
rj = random.randint(0, n2 - 1)

matrix2 = build_matrix(n2, cookie_i=ri, cookie_j=rj, icon='🍪')
print("Матриця:")
print_matrix(matrix2)
find_cookie(matrix2)
```

**Дослідження:**  
1. Використовується `random.randint(0, n-1)` для правильних 0-базових індексів. Іконка може бути в будь-якій клітинці (посередині, на краю, у куті) — алгоритм перевіряє всі клітинки подвійним циклом і знаходить іконку, якщо вона є.  
2. Адреса друкується у 0-базовому та **Візуально** у 1-базовому форматі.
