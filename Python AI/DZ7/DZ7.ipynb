### **Вправа 1.** Дано квадратну матрицю з випадковими числами від 1 до 9. Знайдіть **середнє арифметичне значення чисел** усіх клітинок.

```python
import numpy as np

n = 3  # розмір матриці n x n

#AS_PICTURE = True  # якщо True — матриця як на зображенні
AS_PICTURE = False  # якщо False — випадкова матриця

if AS_PICTURE:
    # матриця як у завданні
    matrix = np.array([[9, 8, 2], [5, 4, 6], [1, 9, 1]])
else:
    # випадкова матриця 1..9 (включно)
    matrix = np.random.randint(1, 10, size=(n, n))

# Середнє арифметичне
mean = matrix.mean()

# Вивід
print('Матриця (', n, 'x', n, '):', sep='')
print(matrix)
print('Середнє арифметичне всіх елементів =', float(mean))
print('\nВідповідь: середнє =', round(float(mean), 2))
```

**Дослідження:**  
1. **`import numpy as np`** — підключаємо бібліотеку `NumPy` для роботи з масивами та матрицями.  
2. **`n = 3`** — задаємо розмір квадратної матриці `n × n`.   
3. `AS_PICTURE = True` — якщо потрібно **взяти матрицю з завдання** — створюємо вручну.  
4. Якщо поставити `AS_PICTURE = False`, створюємо **випадкову** матрицю `n×n` цілих у діапазоні `1..9` (**`np.random.randint(1, 10, size=(n, n))`**, (верхня межа `10` не входить)).   
5. **`matrix.mean()`** — обчислює **середнє арифметичне** всіх елементів матриці.  
6. **Вивід**: друкуємо саму матрицю та її середнє; у рядку **«Відповідь»** показуємо значення з округленням.

### **Вправа 2.** Побудуйте графік функції:  \( y = -2x^2 + 6x - 4 \)

```python
import numpy as np
import matplotlib.pyplot as plt

# Коефіцієнти:
a = -2
b = 6
c = -4

# Область для побудови:
x = np.linspace(-5, 5, 401)

# Наша функція:
y = a*(x**2) + b*x + c  # -2x^2 + 6x -4

# Екстремум (вершина) функції:
xv = -b / (2*a)  # -6/(2*(-2)) = 1.5
yv = a*(xv**2) + b*xv + c  # -2*(1.5**2) + 6*1.5 -4 = 0.5

# Побудова:
plt.figure()
plt.plot(x, y)
plt.axhline(0, linewidth=1)
plt.axvline(0, linewidth=1)
plt.grid(True)
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = -2x^2 + 6x - 4')

# Точки перетину:
# y-перетин (x=0): (0, c), відмічаємо точкою:
plt.scatter([0], [c], marker='o')

# x-перетини (корені), (якщо є) відмічаємо хрестиками:
D = b*b - 4*a*c
if D >= 0:
    x1 = (-b - D**0.5) / (2*a)
    x2 = (-b + D**0.5) / (2*a)
    plt.scatter([x1, x2], [0, 0], marker='x')

# Екстремум (вершина), відмічаємо точкою:
plt.scatter([xv], [yv], marker='o')

# Відображаємо визначені значення з округленням до 2 знаків після коми через round():
print('Відповідь:')
print('  Екстремум (вершина): x_v =', round(float(xv), 2), ', y_v =', round(float(yv), 2))
print('  Перетин з Oy: (0,', c, ')', sep='')
if D >= 0:
    print('  Корені: x1 =', round(float(x1), 2), ', x2 =', round(float(x2), 2))
else:
    print('  Коренів немає')

# Відображаємо графік:
plt.show()
```

**Дослідження:**  
1. **`import numpy as np`, `import matplotlib.pyplot as plt`** — імпортуємо `NumPy` для чисел та `matplotlib` для побудови графіка.  
2. **`a, b, c`** — коефіцієнти квадратичної функції **`y = a*x^2 + b*x + c`**. У нас `a = -2`, `b = 6`, `c = -4`.  
3. **`x = np.linspace(-5, 5, 401)`** — формуємо 401 рівновіддалену точку на відрізку `[-5, 5]` (включно) для плавної побудови.  
4. **`y = a*(x**2) + b*x + c`** — рахуємо значення `y` для кожного `x`.  
5. **Екстремум (вершина)**:  
   - Обчислюємо **`x_v = -b / (2*a)`**, **`y_v = a*(x_v**2) + b*x_v + c`**.  
   - Оскільки `a < 0`, вершина — **максимум**. Позначаємо її **точкою**: `plt.scatter([x_v], [y_v], marker='o')`.  
6. **Осі координат**: `plt.axhline(0, linewidth=1)` (вісь **Ox**) і `plt.axvline(0, linewidth=1)` (вісь **Oy**) - для наочного вигляду.  
7. **Точки перетину**:  
   - З **Oy**: `(0, c)` - `plt.scatter([0], [c], marker='o')` (позначаємо **перетин з віссю Oy** у точці **(0, c)**. Для наших `a, b, c` маємо `(0, -4)`).  
   - З **Ox**: дискримінант **`D = b*b - 4*a*c`**; якщо `D ≥ 0`, обчислюємо корені **`x1, x2`** і позначаємо `plt.scatter([x1, x2], [0, 0], marker='x')` — відмічаємо **перетини з Ox** у точках `(x1, 0)` та `(x2, 0)`.  
8. **Оформлення**: `plt.figure()` — створюємо фігуру, `plt.plot(x, y)` — будуємо криву, `plt.grid(True)` - сітка, підписи осей (`plt.xlabel/plt.ylabel`) і заголовок (`plt.title`), потім відображаємо `plt.show()`.
