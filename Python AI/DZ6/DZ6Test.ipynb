## Примітки:
- **Середнє (mean)** — арифметичне: сума / кількість
- **Медіана (median)** — центральне значення (половина ліворуч, половина праворуч)
- **Мода (mode)** — найчастіше значення
- **Асиметрія**:  
  • правостороння (позитивна): `mode < median < mean` => довший **правий** хвіст  
  • лівостороння (негативна): `mean < median < mode` => довший **лівий** хвіст  
- **σ (сігма)** — стандартне відхилення генеральної сукупності (population SD) у розрахунках відповідає `ddof = 0`
- **s (std)** — стандартне відхилення вибірки (sample SD) у розрахунках відповідає `ddof = 1`
- **Стандартне відхилення (σ)** — це міра типового розкиду даних навколо середнього (`mean`), чим більше `σ` — тим більший розкид, чим менше `σ` — тим дані щільніше біля середнього. Зв’язок із дисперсією: `var = std**2`, `std = sqrt(var)`, де `var` — це дисперсія, середній квадрат відхилень значень від середнього. Вона показує, наскільки розкидані дані, а корінь із неї дає стандартне відхилення в початкових одиницях. Якщо більша `var` — то більший розкид.
- **z-оцінка (z-score)** — **скільки σ** від середнього(`mean`): `z = (x − mean) / σ`

### №2) Середнє арифметичне п’яти чисел 12. Якщо замінити одне число на 20, середнє стає 13. Яке число було замінене?
- А) 5
- Б) 10
- В) 15
- Г) 25

**Відповідь:** **В — 15**

**Дослідження:**  
- Старе середнє: `μ₀ = 12` для 5 чисел ⇒ сума `S₀ = 5·12 = 60`
- Після заміни на `20` нове середнє `μ₁ = 13` ⇒ сума `S₁ = 5·13 = 65`
- Різниця сум: `S₁ − S₀ = (20 − x) = 5` ⇒ `x = 15`
- Перевірка: замінили 15 на 20 → сума зросла на 5, нове середнє справді 13

### №3) Випадкова величина має Дисперсію 25. Яке стандартне відхилення?
- А) 25
- Б) 5
- В) 0.25
- Г) 6.25

**Відповідь:** **Б — 5**

**Дослідження:**  
- Стандартне відхилення — це корінь з дисперсії: `σ = sqrt(var)`. Дисперсія рахує середній квадрат відхилень
- За умовою `var = 25` ⇒ `σ = sqrt(25) = 5`
- Інтерпретація: типове відхилення від середнього становить **5 одиниць**

### №5) У тесті 5 питань, кожне має 4 варіанти. Яка ймовірність вгадати всі відповіді випадково?
- А) 2.0051%
- Б) 0.098%
- В) 0.256%
- Г) 9.822%

**Відповідь:** **Б — 0.098%**

**Дослідження:**  
- Для кожного з 5 питань випадкове вгадування: `P1 = 1/4`
- Усі 5 правильно: `P = (1/4)**5 = 1/4**5 = 1/1024 ≈ 0.0009765625*100% = 0.09765625% ≈ 0.098%`
- Висновок: події незалежні, перемножуємо ймовірності

### №6) Якщо середнє = 10, σ = 2, то значення 14 — це скільки σ(стандартних відхилень) від середнього?
- А) 1
- Б) 2
- В) 3
- Г) 4

**Відповідь:** **Б — 2**

**Дослідження:**  
- z‑оцінка: `z = (x − μ)/σ = (14 − 10)/2 = 2`
- Висновок: отже значення на **2 σ вище** за середнє
- Це 95% всіх даних.

### №8) Яка з наведених статистик є робастною (стійкою до викидів)?
- А) СЕРЕДНЄ
- Б) ДИСПЕРСІЯ
- В) МЕДІАНА
- Г) СТАНДАРТНЕ ВІДХИЛЕННЯ

**Відповідь:** **В — медіана**

**Дослідження:**

- Натомість середнє/дисперсія/σ сильно реагують на викиди

### №9) Монету підкидають 5 разів. Ймовірність, що всі 5 разів випаде аверс:
- А) 0.03125
- Б) 0.0551
- В) 0.09021
- Г) 0.1025

**Відповідь:** **А — 0.03125**

**Дослідження:**  
- 5 підкидань монети, всі аверси: `P = (1/2)**5 = 1/2**5 = 1/32 = 0.03125*100% = 3.125%`
- Висновок: події незалежні, множимо `1/2` п’ять разів

### №10) У шкільному експерименті середній час реакції 0,6С(mean=0.60), медіана 0,58С(median=0.58), мода 0,54С(mode=0.54). Який тип розподілу наймовірніше?
- А) СИМЕТРИЧНИЙ
- Б) ПОЗИТИВНА АСИМЕТРІЯ
- В) НОРМАЛЬНИЙ РОЗПОДІЛ
- Г) ЛІВОСТОРОННЯ АСИМЕТРІЯ

**Відповідь:** **Б — позитивна асиметрія**

**Дослідження:**  
- Дано `mean = 0.60`, `median = 0.58`, `mode = 0.54`
- Співвідношення **`mode < median < mean`** — ознака правосторонньої асиметрії: правий хвіст тягне середнє вправо
- Медіана менше за середнє, але більше за моду — все відповідає **правій асиметрії**

### №13) У наборі 5 елементів, середнє = 10. Якщо додати шосте число 16, яке нове середнє?
- А) 11.5
- Б) 10.5
- В) 11
- Г) 12

**Відповідь:** **В — 11**

**Дослідження:**  
- Початкова сума `S₀ = 5·10 = 50`
- Додаємо 16: `S₁ = 66`. Кількість елементів стала 6
- Нове середнє: `μ₁ = 66/6 = 11`
- Додавання значення **вище** за початкове середнє, підняло середнє вгору

### №15) У якому випадку розподіл даних буде симетричним?
- А) СЕРЕДНЄ > МЕДІАНА
- Б) СЕРЕДНЄ = МЕДІАНА = МОДА
- В) МОДА < МЕДІАНА < СЕРЕДНЄ
- Г) МОДА > МЕДІАНА > СЕРЕДНЄ

**Відповідь:** **Б — середнє = медіана = мода**

**Дослідження:**  
- Для **симетричних** розподілів даних, центральні тенденції збігаються
- При правосторонній - `mode < median < mean`, при лівосторонній - `mean < median < mode`
- Тож рівність трьох показників `mode = median = mean` — індикатор симетричності
