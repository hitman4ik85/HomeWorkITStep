## 6. Монетку підкидають 4 рази. Яка ймовірність, що випаде саме 2 аверси?  

```python
# Імпортуємо біноміальний коефіцієнт C(n, k) з модуля math
from math import comb

def binom(quantity, prob, likely):  # біноміальний розподіл
    """
    Повертає P(X = likely) для X ~ Bin(n=quantity, p=prob).

    Формула:
        P(X = k) = C(n, k) * p^k * (1 - p)^(n - k)
    де:
        n = quantity  — кількість підкидань,
        p = prob      — ймовірність успіху аверсу в одному підкиданні,
        k = likely    — рівно стільки успіхів аверсів, скільки нас цікавить.
    """
    return comb(quantity, likely) * (prob ** likely) * ((1 - prob) ** (quantity - likely))

# 4 підкидання (n=4), p=0.5 (чесна монета), шукаємо P(X=2)
p = binom(4, 0.5, 2)
print(f"P(X=2) = {p:.3f}")        # у частках
print(f"Outcome: {p*100:.1f} %")  # у відсотках
```

**Варіанти:**  
- **А.** 0.125  
- **Б.** 0.375  
- **В.** 0.488  
- **Г.** 0.285  

**Відповідь:** **Б. 0.375**  

**Пояснення:**  
- Для **4 підкидань чесної монети** ймовірність побачити **рівно 2 аверси** дорівнює **0.375 = 37.5%**.  
- Тобто, якщо багато разів повторювати **серії по 4 підкидання**, то **приблизно у 37.5% серій** ми спостерігатимемо **рівно два аверси**.  
- Якщо потрібно інші `k`, формула залишається тією самою: `P(X=k)=C(n,k)*(p^k)*(1-p)^(n-k)`.



## 7. Дано набір даних: 1, 4, 8, 1, 6, 50. Що буде медіаною?  

```python

# Вхідні дані (невпорядкований список)
data = [1, 4, 8, 1, 6, 50]

# 1) Сортуємо: створюється НОВИЙ список; data не змінюється
s = sorted(data)   # s = [1, 1, 4, 6, 8, 50]

# 2) Для парної кількості елементів n=6:
#    m = 3 → центральні індекси: left=2 і right=3 (нумерація з 0)
left_i = 2
right_i = 3

# 3) Обчислюємо медіану як середнє двох центральних
med = (s[left_i] + s[right_i]) / 2  # (4 + 6) / 2 = 5.0

# 4) Вивід результатів
print('Відсортовано:', s)
print('Медіана =', med)
```

**Варіанти:**  
- **А.** 4  
- **Б.** 5  
- **В.** 6  
- **Г.** 6.5  

**Відповідь:** **Б. 5**  

**Пояснення:**  
- Відсортуємо дані: `1, 1, 4, 6, 8, 50`.  
- Кількість елементів парна (`n=6`), тому медіана — **середнє двох центральних** значень.  
- Центральні значення: `4` і `6`; медіана = `(4 + 6) / 2 = 5`.  
- Зверніть увагу: великий елемент `50` (викид) **майже не впливає** на медіану, на відміну від середнього.

## 10. У супермаркеті 20% покупців реагують на знижку. Яка ймовірність, що серед 15 відвідувачив супермаркету, придбають товар хоча б 5?  

```python
# Імпортуємо біноміальний коефіцієнт C(n, k) з модуля math
from math import comb

def binom(quantity, prob, likely):  # біноміальний розподіл (точна ймовірність)
    """
    Повертає P(X = likely) для X ~ Bin(n=quantity, p=prob).

    Формула:
        P(X = k) = C(n, k) * p^k * (1 - p)^(n - k)
    де:
        n = quantity  — кількість випробувань покупців,
        p = prob      — ймовірність успіху покупки в одному випробуванні,
        k = likely    — рівно стільки успіхів покупок, скільки нас цікавить.
    """
    return comb(quantity, likely) * (prob ** likely) * ((1 - prob) ** (quantity - likely))

# Умова: p=0.20, n=15. Потрібно: P(X ≥ 5) — «хоча б 5 покупок»
n = 15
p = 0.20

p_tail = sum(binom(n, p, k) for k in range(5, n + 1))  # сума точних імовірностей від 5 до 15
print(f"P(X >= 5) = {p_tail:.6f}")          # у частках
print(f"Outcome: {p_tail*100:.2f} %")       # у відсотках
```

**Відповідь:** **≈ 16.42%**.  

**Дослідження:**  
- `binom(n, p, k)` реалізує формулу біноміального розподілу  
   `P(X = k) = C(n, k) * p^k * (1 - p)^(n - k)` — **точна** ймовірність рівно k успіхів.
- «**Хоча б 5**» означає **сума** точних імовірностей:  
  `P(X ≥ 5) = Σ_{k=5..15} C(15,k) * 0.2^k * 0.8^(15-k)`.
- Код обчислює цю суму через нашу функцію `binom(...)`.

**Пояснення:**  
- Отримуємо: **P(X ≥ 5) ≈ 0.164234 → 16.42%**.  
- Це означає: якщо багато разів повторювати **серії по 15 покупців** із шансом покупки **20%** кожного, то **приблизно у 16–17% серій** ми побачимо **не менше 5 покупок**.  
- Для задачі «**рівно 5**» використовуємо **одну** формулу `P(X=5) = C(15,5)*0.2^5*0.8^10`, `P(X=5) ≈ 0.10318 → 10.32%`.  
- Для «**хоча б 5**» — **суму** від 5 до 15.
