## Кореляція Пірсона (між temp та hurricanes)

```python
import pandas as pd
from dataclasses import dataclass
from scipy.stats import pearsonr

df = pd.read_csv('DZ8.csv') # завантаження історичних даних про температуру та кількість ураганів у вигляді DataFrame

@dataclass(frozen=True)
class Pearson:
  corr: float # коефіцієнт кореляції Пірсона для двох змінних TEMP та HURRICANES
  p_value: float # p-value для перевірки значущості кореляції
  n: int = len(df['TEMP']) # кількість спостережень (рядків) у вибірці

def pearson(x, y) -> Pearson:
  corr, p = pearsonr(x, y) # обчислення коефіцієнта кореляції Пірсона та p-value
  return Pearson(corr=float(corr), p_value=float(p))

let = pearson(df['TEMP'], df['HURRICANES']) # беремо дані про температуру та кількість ураганів за той самий рік та обчислюємо кореляцію Пірсона
 
print(f'\nТаблиця, розмірністю {df.shape}:\n {df.head(10)}')
print(f"""\nЗначення коефіцієнта кореляції Пірсона та p-value для ознак TEMP та HURRICANES:
  r = {let.corr:.4f}
  p_value = {let.p_value:.14f}
  r^2 = {let.corr**2:.2f}
""".rstrip())
```

___
**Дослідження (опис коду):**  
1. **Імпорти**: `import pandas as pd` та **`from scipy.stats import pearsonr`**, `from dataclasses import dataclass`. 
2. **Читання даних**: `df = pd.read_csv('DZ8.csv')`. У таблиці є стовпці **`YEAR`**, **`TEMP`**, **`HURRICANES`**.  
3. **Оголошуємо структуру результату** через `@dataclass(Pearson)` з полями:  
   - `corr`, `r` — це коефіцієнт кореляції Пірсона між двома змінними (у нашому випадку TEMP і HURRICANES). Діапазон: від −1 до 1; 
      - знак `+` → прямий зв’язок (зростає одне → має тенденцію зростати інше);
      - знак `−` → обернений зв’язок;
      - за `0` → лінійного зв’язку немає;
      - за `|r|` → оцінюємо силу зв’язку (чим ближче до 1, тим сильніше). 
   - `p_value` — рівень значущості для `r`, статистично значуща ознака, на скільки сильно наше поточне значення може відхилятися від центра нашого дослідження. Ймовірність отримати не менший за абсолютною величиною `r` за умови, що реального зв’язку немає (нульова гіпотеза).  
      - типове правило: якщо `p-value < 0.05`, то вважаємо зв’язок статистично значущим;
      - якщо `p-value ≥ 0.05`, то доказів недостатньо, щоб стверджувати про значущий лінійний зв’язок.  
   - А `let` — це об’єкт, у якому ми зберігаємо ці значення й звідки їх зручно друкуємо.
   - `n` — кількість спостережень (беремо довжину колонки).  
4. **Функція** `pearson(x, y)`: викликає `pearsonr(x, y)`, отримує `(corr, p)` і повертає `Pearson(corr=..., p_value=...)`.  
5. **Обчислення**: викликаємо `pearson(df['TEMP'], df['HURRICANES'])` і **друкуємо**.  
___
**Інтерпретація (відповідь):**  
- Отримали: **`r = 0.8122`**, **`p_value = 0.0043`**.  
- **Напрям зв’язку**: `r > 0` → **прямий лінійний зв'язок** (за зростанням температури `TEMP` спостерігається тенденція до збільшення кількості ураганів `HURRICANES` за той самий період `YEAR`).  
- **Сила зв’язку**: `|r| ≈ 0.81` → це **сильний** лінійний зв’язок (лежить у діапазоні 0.7–0.9, що прийнято казати, як сильний лінійний зв’язок).  
- **Статистична значущість**: `p_value ≈ 0.0043 < 0.05` → зв’язок **статистично значущий** (за нульової гіпотези **лінійного зв’язку немає** імовірність побачити таке, або сильніше `|r|` — `~0.43%`).  
- **Коефіцієнт детермінації**: `r^2≈0.66` — `~66%` лінійної варіації `HURRICANES` узгоджується з варіацією `TEMP` у нашій вибірці. Залишок `~34%` припадає на інші фактори.  
- **Розмір вибірки (`n`)**: береться як довжина колонки (`len(df['TEMP'])` у нашому `@dataclass`). Чим більше `n` — стабільніша оцінка `r` і менше `p_value` за тієї ж сили зв’язку.

**Висновок:** 
- За наявними даними існує **сильний, прямий і статистично значущий** лінійний зв’язок між температурою поверхні океану (`TEMP`) та кількістю ураганів (`HURRICANES`) за той самий період(`YEAR`).  
- Кореляція показує **лінійну асоціацію**, але **не доводить причинно-наслідковий зв’язок**. Сильний додатний `r` узгоджується з гіпотезою, що вищі температури поверхні океану часто супроводжуються більшою кількістю ураганів у тому ж році. Це може бути корисно для спостережного моніторингу чи прогнозування (але не як доказ причинності).
