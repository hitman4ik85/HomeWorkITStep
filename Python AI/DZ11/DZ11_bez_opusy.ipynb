```python
import pandas as pd
```

```python
df = pd.read_csv('scores.csv')
df
```

```python
for i in ['math', 'economics', 'ukrainian']:
    df[f"{i}_mark"] = pd.cut(
        df[i],
        bins=[0, 34, 59, 63, 73, 81, 89, 100],
        labels=['F', 'FX', 'E', 'D', 'C', 'B', 'A'],
        include_lowest=True
    )
df
```

```python
updated_df = df.melt(
    id_vars="id",
    value_vars=['math_mark', 'economics_mark', 'ukrainian_mark'],
    var_name="subject",
    value_name="mark"
)
updated_df
```

```python
updated_df['subject'] = updated_df['subject'].str.replace("_mark", "")
```

```python
table = pd.crosstab(updated_df["mark"], updated_df["subject"])
table
```

```python
from math import comb

def binom(quantity, prob, likely):  # біноміальний розподіл
    return comb(quantity, likely) * (prob ** likely) * ((1 - prob) ** (quantity - likely))

# 15 подач, ймовірність ейсу 0.18, хочемо рівно 4 ейси
p = binom(15, 0.18, 4) * 100

print(f"Outcome: {p:.1f} %")
```
