## Вправа 1 — лінійна модель у dataclass

```python
from dataclasses import dataclass

@dataclass
class Line:
    slope: float                      # нахил (коефіцієнт при x)
    intercept: int | float = 1.       # перетин з віссю OY (при x=0)

    def model(self, x: float) -> float:
        return self.slope * x + self.intercept

line = Line(5, 8)
line.model(2.28)
```

**Дослідження:**  
`@dataclass` Це декоратор з модуля dataclasses, який створює клас даних `Line` з полями `slope` (нахил) та `intercept` (перетин з OY).  
- Поля класу:  
  - `slope: float` — **нахил** прямої на графіку (на скільки зростає `y`, коли `x` збільшується на 1).  
  - `intercept: int | float = 1.` — **перетин з Ordinat Y**, значення `y`, коли `x = 0`. Значення **1.0** — це **дефолтне**, але ми можемо передати інше при створенні об’єкта.  
- Метод `model(self, x: float) -> float` реалізує формулу **`y = slope * x + intercept`**.  
- Приклад: датакласс `line = Line(5, 8)` визначає коефіцієнти `slope = 5` та `intercept = 8`.  
  Виклик `line.model(2.28)` → де `х` це поточне значення нашої незалежної ознаки, яке надходить в наш датакласс ззовні. Підстановка `x = 2.28`:  
  `y = 5 * 2.28 + 8 = 11.4 + 8 = 19.4`.

## Вправа 2 — клас з __call__(метод який викликає самого себе) для StandardScaler

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class StandardScaler:
    mu: float     # середнє значення генеральної сукупності (мат. сподівання)(mean)
    sigma: float  # стандартне відхилення (міра розкиду)(std)

    def __call__(self, xi: float) -> float:
        # стандартна оцінка: (xi - mu) / sigma
        if self.sigma == 0:
            return 0  # уникаємо ділення на нуль
        return (xi - self.mu) / self.sigma

# приклади використання
scaler = StandardScaler(mu=10., sigma=2.5)
(
    scaler(12.),   # (12 - 10) / 2.5 = 0.8
    scaler(10.),   # (10 - 10) / 2.5 = 0
    scaler(7.5)     # (7.5 - 10) / 2.5 = -1.0
)
```

**Дослідження:**  
- Зберігаємо параметри **розподілу**: `mu` (середнє (математичне сподівання) по вибірці) та `sigma` (стандартне відхилення (міра розкиду навколо mu)).  
- Метод **`__call__`** робить екземпляр **викличним**: тобто `scaler(xi)` працює як функція.  
- Формула **стандартизації** (z-score): `z = (xi − mu) / sigma`.  
  - якщо `z ≈ 0` → `xi` близько до середнього;  
  - `z = 1` → на одну σ **вище** середнього;  
  - `z = −1` → на одну σ **нижче** середнього.  
- Обробка особливого випадку: якщо `sigma == 0` (усі значення однакові), ділення заборонене — повертаємо **0**.  
- Приклади для `mu=10.0`, `sigma=2.5`:  
  - `scaler(12.0)` → `(12 − 10)/2.5 = **0.8**` (трохи вище середнього),  
  - `scaler(10.0)` → `0` (рівно середнє),  
  - `scaler(7.5)` → `(7.5 − 10)/2.5 = **−1.0**` (на одну σ нижче середнього).  

```python
#робив викладач на парі
from dataclasses import dataclass, field

@dataclass(frozen=True) #перетворює наш клас на структуру даних
class Standard:
  mean: float
  std: float = field(default=1.)

  def __call__(self, i):
    if self.std == 0:
            return 0
    return (i - self.mean) / self.std
  
standard = Standard(50, 5)
standard(25)
```
