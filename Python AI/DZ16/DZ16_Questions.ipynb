```python
# 1) Створення списку квадратів чисел від 0 до 9
squares = [a**2 for a in range(10)]
squares
```

### Пояснення:
- Тут використано **list comprehension** — це короткий і читабельний спосіб створити список.
- `range(10)` генерує числа від 0 до 9, а `a**2` підносить кожне число до квадрату.
- У результаті отримаємо список `[0, 1, 4, 9, 16, 25, 36, 49, 64, 81]`.
- На співбесіді я б підкреслив, що list comprehension зазвичай виглядає охайніше й часто працює трохи швидше, ніж класичний цикл `for` з `.append()`.

```python
# 4) Активаційна функція Sigmoid у PyTorch
sigmoid = torch.nn.Sigmoid()
x = torch.tensor([-1.0, 0.0, 1.0])
sigmoid(x)
```

### Пояснення:
- `torch.nn.Sigmoid()` — це активаційна функція, яка перетворює значення в діапазон (0, 1).
- Часто використовується в **бінарних класифікаторах** на вихідному шарі, де результат інтерпретується як ймовірність класу.
- Формула: `1 / (1 + exp(−x))`.
- Для багатокласових задач зазвичай використовують `softmax`, а `sigmoid` — для задач з одним виходом або multi-label.

```python
# 5) Обчислення градієнтів через backward()
w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0)
y_pred = w * x
loss = (y_pred - 10)**2
loss.backward()
w.grad
```

### Пояснення:
- `loss.backward()` запускає **backpropagation** — обчислення градієнтів для всіх параметрів з `requires_grad=True`.
- Після цього градієнти потрапляють у `.grad` кожного параметра (наприклад, `w.grad`).
- Це ключовий крок навчання нейромереж: далі оптимізатор використовує ці градієнти для оновлення ваг.
- Я завжди згадую типову схему: `zero_grad()` → forward → `loss.backward()` → `optim.step()`.

```python
# 10) Ймовірності класів у sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X = [[0], [1], [2], [3]]
y = [0, 0, 1, 1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
clf = LogisticRegression()
clf.fit(X_train, y_train)
clf.predict_proba(X_test)
```

### Пояснення:
- `clf.predict_proba(X_test)` повертає **ймовірності класів** для кожного прикладу.
- Для бінарної класифікації це дві колонки: P(клас=0) та P(клас=1).
- Це потрібно для побудови ROC/PR-кривих, налаштування порога рішення та інтерпретації впевненості моделі.
- Я підкреслюю різницю між `predict()` (класи) та `predict_proba()` (ймовірності).

```python
# 18) Softmax для логітів по виміру класів
logits = torch.randn(2, 3)
probs = torch.softmax(logits, dim=1)
probs, probs.sum(dim=1)
```

### Пояснення:
- `torch.softmax(logits, dim=1)` перетворює логіти на розподіл ймовірностей по класах.
- Сума по вказаній осі дорівнює 1.
- Використовується в багатокласових класифікаторах на вихідному шарі.
- Часто йде в парі з `CrossEntropyLoss`, яка всередині працює з логітами.

```python
# 19) Крок оптимізатора — оновлення ваг моделі
model_simple = torch.nn.Linear(1, 1)
optimizer = torch.optim.SGD(model_simple.parameters(), lr=0.1)
x = torch.tensor([[1.0]])
y_true = torch.tensor([[2.0]])
y_pred = model_simple(x)
loss = (y_pred - y_true)**2
loss.backward()
optimizer.step()
```

### Пояснення:
- `optimizer.step()` оновлює параметри моделі, використовуючи вже обчислені градієнти.
- Це реалізація конкретного алгоритму оптимізації (SGD, Adam тощо).
- Завжди йде після `loss.backward()` і зазвичай після `optimizer.zero_grad()`.
- Це серце процесу навчання: саме тут змінюються ваги моделі.

```python
# 22) Генератор та next()
obj = (a + 1 for a in range(5))
first = next(obj)
second = next(obj)
first, second
```

### Пояснення:
- `(a+1 for a in range(5))` — генераторний вираз, який обчислює елементи по одному, коли вони потрібні.
- `next(obj)` повертає наступний елемент і зсуває «вказівник» усередині генератора.
- Генератори зручні, коли не потрібно тримати всі дані в пам'яті, наприклад, при стрімінгу даних.
- На співбесіді це показує розуміння ітераторів і роботи протоколу `next()` у Python.
