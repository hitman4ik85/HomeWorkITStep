# DZ16_Vprava — задача про національний парк (коло)

Національний парк має форму кола з центром у точці **C(6; 2)** та радіусом **R = 4 км**.

## А) Опишіть рівняння межі парку

Межа парку — це коло з центром у точці `C(6; 2)` і радіусом `R = 4`.

Загальне рівняння кола з центром у точці `C(x₀; y₀)` та радіусом `R` має вигляд:

- `(x − x₀)² + (y − y₀)² = R²`

Підставимо координати центру `x₀ = 6`, `y₀ = 2` та радіус `R = 4`:

- `(x − 6)² + (y − 2)² = 4²`
- `(x − 6)² + (y − 2)² = 16`

**Відповідь (А):** рівняння межі парку:

- **(x − 6)² + (y − 2)² = 16**

## Б) Чи знаходиться лавка в точці (2; 5) біля межі парку?

Лавка має координати точки `P(2; 5)`.

Щоб зрозуміти, де вона розташована відносно парку, порівняємо **відстань від центру кола до точки** з радіусом кола.

### 1. Обчислюємо відстань CP

Центр парку: `C(6; 2)`

Лавка: `P(2; 5)`

Різниця координат:
- по x: `2 − 6 = −4`
- по y: `5 − 2 = 3`

Квадрат відстані `CP²`:
- `CP² = (−4)² + 3² = 16 + 9 = 25`
Тоді відстань:
- `CP = √25 = 5`

### 2. Порівнюємо з радіусом

- Радіус парку: `R = 4`
- Відстань до лавки: `CP = 5`

Маємо: `5 > 4`, отже точка **P** лежить **поза колом** (за межами парку).

**Висновок (Б):**

- Лавка **не знаходиться** на межі парку і не всередині нього — вона на відстані 5 км від центру, тобто на **1 км далі**, ніж радіус парку.

## В) Вулиця задана рівнянням x + y = 12. Чи перетинає вона парк? Якщо так, чи потрібно будувати доріжку?

Маємо коло з рівнянням:
- `(x − 6)² + (y − 2)² = 16`

та пряму (вулицю) з рівнянням:
- `x + y = 12`, або еквівалентно `x + y − 12 = 0`.

Щоб дізнатися, чи пряма перетинає коло, обчислимо **відстань від центру кола до прямої** і порівняємо її з радіусом.

### 1. Формула відстані від точки до прямої

Для прямої `Ax + By + C = 0` і точки `C(x₀; y₀)` відстань:

- `d = |A·x₀ + B·y₀ + C| / √(A² + B²)`

У нашому випадку: `x + y − 12 = 0`, тобто `A = 1`, `B = 1`, `C = −12`.

Центр кола: `C(6; 2)`

Підставляємо в чисельник:
- `|1·6 + 1·2 − 12| = |6 + 2 − 12| = |−4| = 4`

Знаменник:
- `√(1² + 1²) = √(1 + 1) = √2`

Отже відстань від центру до прямої:
- `d = 4 / √2 = 2·√2 ≈ 2.83`

### 2. Порівнюємо з радіусом кола

- Радіус парку: `R = 4`
- Відстань до вулиці: `d ≈ 2.83`

Маємо `d < R`, тобто `2.83 < 4`, отже **пряма перетинає коло** (вулиця проходить через територію парку).

### Висновок (В)

- Так, вулиця `x + y = 12` **перетинає парк**.
- Оскільки дорога вже заходить на територію парку, **додаткову доріжку будувати не потрібно** — доступ до парку вже є безпосередньо з цієї вулиці.
