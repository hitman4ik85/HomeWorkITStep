# DZ3 — Щоденник погоди


```python
import pandas as pd
```

```python
data = {
    'Date': pd.date_range('2024-09-01', periods=10, freq='D'),
    'Temp': [18, 21, 19, 23, 24, 17, 20, 22, 25, 16],
    'State': ['☀️', '🌤️', '☁️', '☀️', '⛅', '🌧️', '☀️', '☀️', '🌧️', '☀️'],
    'Wind': [3.2, 2.8, 5.5, 4.0, 6.1, 3.5, 7.4, 2.0, 3.2, 4.8]
}
df = pd.DataFrame(data)
df
```

**Дослідження:**
- Створюємо DataFrame `df` з назвами стовпчиків: **Date, Temp, State, Wind**.
- `State` містить **EMOJI** стану погоди (☀️, 🌤️, ☁️, ⛅, 🌧️).
- Виводимо `df`, щоб бачити весь щоденник погоди.

```python
mean_temp = df['Temp'].mean()
mean_temp
```

**Дослідження:**
- `mean()` обчислює **середню температуру повітря** за всі дні з колонки `Temp`.
- Результат зберігаємо в `mean_temp` і відображаємо останнім рядком клітинки.

```python
count_sunny = df.query("State == '☀️'").shape[0]
#count_sunny = (df['State'] == '☀️').sum()
count_sunny
```

**Дослідження:**  
1:
- Використовуємо запит `df.query("State == '☀️'")` для **фільтрації** лише сонячних днів.  
- Властивість `.shape[0]` повертає **кількість рядків** після фільтра — це і є **кількість сонячних днів**.  

2:
- Умова `df['State'] == '☀️'` створює булевий стовпчик True/False.  
- Функція `sum()` рахує кількість True — це і буде **кількість сонячних днів**.

```python
windy_day = df[df['Wind'] == df['Wind'].max()]
windy_day
windy_date = list(windy_day['Date'])[0]
windy_date
```

**Дослідження (ADVANCED):**
- `df['Wind'].max()` — знаходимо **максимальну швидкість вітру**.
- `df[df['Wind'] == ...]` — відбираємо рядок(и) з цим значенням.
- Беремо стовпчик `Date` і перший елемент `list(...)[0]` — **дату найшвидшого вітру**.
